```python
mtb.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
mtb.title("Correlation Heatmap - Seoul Bike Dataset")
mtb.show()
```

The Above Heat map describes that **Red-strong correlation(+ve) ,Blue-no correlation(-ve) and remaining colors shows some relations,the changing colors shows the how strong the relation between two variables.** 
Examples:
1.Temperature(0.54)-that is temp is correlated with rented bikes
2.Humidity -Negative correlation with rented bikes
3.visibility -shows less opposite correlation with humidity
4.visibility and solar radiation--no correlation
*This Heatmap shows us new insights to move further*
**-ve(opposite correlation),0(No correlation),+ve(less or more correlation)**

```python
sns.set_style("whitegrid")
custom_palette = sns.color_palette("husl", 10)
sns.set_palette(custom_palette)
sns.pairplot(df, hue="Seasons")
```
