1. Difference between AI, ML, and Deep Learning (with examples)

Artificial Intelligence(AI):
Artificial intelligence is the machine that mimic the human intelligence.

Eg:A chess-playing system like IBM DEEP BLUE

Machine Learning(ML):
ML is a subset of AI where system learns patterns from data
Eg:A spam email classifier trained on labeled emails.

Deep Learning(DL):
DL is a subset of ML that uses multi-layered neural networks to learn complex patterns.
Eg:Face recognition


2. Perceptron: Definition and Mathematical Equation

perceptron is the simplest artificial neuron used for binary classification.

It computes a weighted sum of inputs and passes it through an activation function.


y=f(w.x+b)


3. Activation Function: Purpose + Sigmoid, ReLU, Tanh

An activation function introduces non-linearity into the neural network.

Purpose:

1.Allows networks to learn complex patterns

2.Decides whether a neuron should be activated

3.Helps in gradient-based learning


(a) Sigmoid:

f(x)=1/1+e^-x

(b) ReLU (Rectified Linear Unit):

f(x)=max(0,x)

(c)Tanh:

f(x)=e^x-e^-x/e^x+e^-x



4. Architecture of a Basic Artificial Neural Network (ANN)
Input Layer:

Takes raw features (e.g., pixel values)

Hidden Layer(s):

Performs intermediate computations

Learns complex patterns

Output Layer:

Produces final prediction

Weights:

Strength of connections between neurons

Bias:

Adjusts neuron activation threshold

Forward Propagation:
Data flows from input to hidden layers to output using:

z=Wx+b
y=f(z)


5. Backpropagation and Gradient Descent

Backpropagation:
An algorithm to update weights by propagating error backward through the network.

Role of Chain Rule:
The chain rule is used to compute how each weight contributes to the final error:

New Parameter = Current Parameter - (Learning Rate × Gradient of the Loss Function)

Gradient Descent is Used because-

<-It minimizes the loss function

<-Updates weights in the direction of steepest descent

6. Batch vs Stochastic vs Mini-batch Gradient Descent

Batch GD--->Entire dataset used
SGD-->One sample used
Mini_batch GD-->Small batch(eg.,32,64)

7. Overfitting and Techniques to Prevent It

Overfitting:
When a model performs well on training data but poorly on unseen data.

Techniques to prevent overfitting:

1.Dropout – randomly disables neurons during training

2.Regularization (L1/L2) – penalizes large weights

3.Early Stopping – stop training when validation loss increases

4.Data Augmentation – increase training data variety

5.Cross-validation

8. What is a CNN? Why Preferred for Images?

-->Convolutional Neural Network (CNN):
A deep learning model designed to process grid-like data such as images.

-->Why CNNs are preferred for image data:

1.Automatically extract features

2.Capture spatial relationships

3.Fewer parameters than fully connected networks

4.Translation invariant

Real-world use:
Used in computer vision systems by companies like Google and Meta for image recognition.

9. Components of CNNs

(a) Convolution Layer:
Applies filters to detect features (edges, textures)

(b) Filter/Kernel:
Small matrix (e.g., 3×3) that slides over the image

(c) Stride:
Step size of the filter movement

(d) Padding:
Adds extra pixels (usually zeros) around the image to preserve size

(e) Pooling Layer:
Reduces spatial size (e.g., Max Pooling)

10. Output Dimension Calculation (CNN)

Given:

Input = 32 × 32 × 3

Filter size = 3 × 3

Number of filters = 10

Stride = 1

Padding = 0

Formula:

Output Size=N-F+2P/S+1

Output Height=32-3+0/1+1=30

Output Width=30

Final Output Dimension=30*30*10
