```python

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image



image_path = r"C:\ML_PROJECT\Deep_Learning\army-day.jpg"
img = Image.open(image_path)
img = img.resize((256,256))  


```

```python

plt.figure()
plt.title("Original Image")
plt.imshow(img)
plt.axis("off")
plt.show()
```

```python

img_gray = np.mean(np.array(img), axis=2)

plt.figure()
plt.title("Grayscale Image")
plt.imshow(img_gray)
plt.axis("off")
plt.show()
```

```python

def convolution(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape
    output = np.zeros((h - kh + 1, w - kw + 1))
    
    for i in range(h - kh + 1):
        for j in range(w - kw + 1):
            region = image[i:i+kh, j:j+kw]
            output[i, j] = np.sum(region * kernel)
            
    return output
```

```python

kernel = np.array([[1, 0, -1],
                   [1, 0, -1],
                   [1, 0, -1]])

conv_output = convolution(img_gray, kernel)

plt.figure()
plt.title("After Convolution (Feature Map)")
plt.imshow(conv_output)
plt.axis("off")
plt.show()
```

```python

relu_output = np.maximum(0, conv_output)

plt.figure()
plt.title("After ReLU Activation")
plt.imshow(relu_output)
plt.axis("off")
plt.show()
```

```python

def max_pooling(image, size=2):
    h, w = image.shape
    output = np.zeros((h//size, w//size))
    
    for i in range(0, h - size + 1, size):
        for j in range(0, w - size + 1, size):
            region = image[i:i+size, j:j+size]
            output[i//size, j//size] = np.max(region)
            
    return output
```

```python
pool_output = max_pooling(relu_output, 2)

plt.figure()
plt.title("After Max Pooling")
plt.imshow(pool_output)
plt.axis("off")
plt.show()

print("Pipeline Complete: Original → Grayscale → Convolution → ReLU → Pooling")
```
